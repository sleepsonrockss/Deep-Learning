# Linear Algebra operations

```python
x_1 = tf.constant([[1,2,0], [3,5,-1]])
x_2 = tf.constant([[3,4],[7,2],[9,2]]) 

tf.linalg.matmul(
    x_1,
    x_2,
    transpose_a=False,
    transpose_b=False,
    adjoint_a=False,
    adjoint_b=False,
    a_is_sparse=False,
    b_is_sparse=False,
    output_type=None,
    grad_a=False,
    grad_b=False,
    name=None
)
```

```python
print(x_1@x_2) # '@' means matrix multiplication, '*' means element wise multiplication
```

```python
tf.transpose(x_1)
```

```python
x_1 = tf.constant([[[1,2],
                    [2,3]],

                    [[4,5],
                    [6,8]],

                    [[1,2],
                    [2,3]]
                    ])

x_2 = tf.constant([[[1,2],
                    [2,3]],

                    [[1,2],
                    [2,3]],

                    [[4,5],
                     [6,8]]])
print(x_1.shape, x_2.shape)
print(tf.linalg.matmul(x_1, x_2))
```

# EinSum Operator

```python
A = np.array([[2,6,5,2], 
              [9,6,4,0],
              [4,7,4,8]])

B = np.array([[2,9,0,3,0],
              [4,5,2,1,4],
              [1,2,5,6,3],
              [5,7,3,5,2]])

print(A.shape, B.shape)
print(np.einsum('ij,jk -> ik', A,B))
```

```python
A = np.array([[2,6,5,2], 
              [9,6,4,0],
              [4,7,4,8]])

B = np.array([[2,9,0,3],
              [4,5,2,1],
              [1,2,5,6]])

print(A.shape, B.shape)
print(np.einsum('ij,ij -> ij', A,B)) #matrix element wise multiplication
```

```python
A = np.array([[2,6,5,2], 
              [9,6,4,0],
              [4,7,4,8]])
print(np.einsum('ij -> ji', A)) #getting the transpose
```

```python
A = tf.constant([[[1,2,3],
                [2,3,4],
                [5,6,7]],

                [[1,2,3],
                [2,3,4],
                [5,6,7]]])

B = tf.constant([[[1,2,3],
                [2,3,4],
                [5,6,7]],

                [[1,2,3],
                [2,3,4],
                [5,6,7]]])

print(np.einsum('bij,bjk -> bik', A,B)) # this is matrix multiplication
```

```python
A = tf.constant([[[1,2,3],
                [2,3,4],
                [5,6,7]],

                [[1,2,3],
                [2,3,4],
                [5,6,7]]])

print(np.sum(A))
print(np.einsum('bij ->', A)) # no output mean that we are summing up all the possible values
```

```python
A = np.array([[2,6,5,2], 
              [9,6,4,0],
              [4,7,4,8]])
print(np.sum(A, axis = 1)) # sum across columns
print(np.einsum('ij->j', A))
print('_______')
print(np.einsum('ij->i', A))
```

# Padding

```python
t = tf.constant([[1, 2, 3],
                [4, 5, 6]])
paddings = tf.constant([[1, 1], [2, 2]]) #1 row above the 1 row below, 2 columns to the left and 2 columns to the right
# 'constant_values' is 0.
# rank of 't' is 2.
tf.pad(t, paddings, "CONSTANT", constant_values=0)
```

```python
params = tf.constant(['p0', 'p1', 'p2', 'p3', 'p4', 'p5'])
print(tf.gather(params, [1,2,3])) #like indexing
print(tf.gather(params, tf.range(1,4)))
print(tf.gather(params, [0,3,5]))
```

```python
indices = [[1]],
params = [['a', 'b', 'c'], 
          ['d', 'e', 'f']]

print(tf.gather_nd(params, indices))

indices = [[1,2]]
print(tf.gather_nd(params, indices))

indices = [[0],[1]]
print(tf.gather_nd(params, indices))
```

```python
indices = [[0,1],[1,1]]
params = [
            [['a0', 'b0'],
            ['c0', 'd0']
            ],

            [['a1', 'b1'],
            ['c1', 'd1']
            ]
        ]
print(tf.gather_nd(params, indices)) #take the 0th element and it's 1st index
indices = [[0,1,0]] #0th element then the 1st row and then the 0th column
print(tf.gather_nd(params, indices))
```

```python
indices = [[[0, 1], 
            [1, 0]], 

            [[0, 0], 
             [1, 1]]]

params = [[['a0', 'b0'], 
           ['c0', 'd0']],

            [['a1', 'b1'], 
             ['c1', 'd1']]]

print(tf.gather_nd(params, indices))
```

```python
indices = [[0, 1], 
            [1, 0]]

params = [[['a0', 'b0'], 
           ['c0', 'd0']],

            [['a1', 'b1'], 
            ['c1', 'd1']]]

print(tf.gather_nd(params, indices, batch_dims=1)) #batchdims = 1 means that it is batch aware
# Notice that the indices list is 2D and it's 0th element corresponds to the 0th batch of params and the 1st element corresponds to the 1st batch of params
```
