```python
import tensorflow as tf
```

# Tensor Dimensions

```python
tensor_zero_d = tf.constant(4) #0 d tensor
print(tensor_zero_d)
```

```python
tensor_one_d = tf.constant([2,0,-1], dtype = tf.float32)
casted_tensor_one_d = tf.cast(tensor_one_d, dtype = tf.int64)
bool_tensor_one_d = tf.cast(tensor_one_d, dtype = tf.bool)
print(tensor_one_d)
print(casted_tensor_one_d)
print(bool_tensor_one_d)
```

```python
tensor_bool = tf.constant([True, False, True])
print(tensor_bool)
```

```python
tensor_two_d = tf.constant([[1,2,3],
                             [2,3,4],
                             [5,6,7]])
print(tensor_two_d)
```

```python
tensor_three_d = tf.constant([[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]])
print(tensor_three_d.shape)
tensor_three_d_2 = tf.constant([[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]])
print(tensor_three_d_2.ndim)
```

```python
print(tensor_three_d.ndim) # Dimension of our tensor
```

# 4D Tensor are 3D tensors stacked 

```python
tensor_four_d = tf.constant([[[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]]
                             ,
                             [[[15,24,35],
                             [25,33,41],
                             [45,66,73]],

                             [[16,26,37],
                             [28,39,40],
                             [54,65,78]]]
                             ])
print(tensor_four_d.ndim)
print(tensor_four_d.shape)
```

## Convert a numpy array to a tensor

```python
import numpy as np
np_array = np.array([1,2,4])
print(np_array)
```

```python
converted_tensor = tf.convert_to_tensor(np_array)
print(converted_tensor)
```

## Eye Tensor

```python
eye_tensor = tf.eye(
    3,
    num_columns=None,
    batch_shape=[2,4],
    dtype=tf.dtypes.int32,
    name=None
)
print(eye_tensor)
```

```python
print(3 * eye_tensor)
```
