# Tensor Dimensions

```python
tensor_zero_d = tf.constant(4) #0 d tensor
print(tensor_zero_d)
```

```python
tensor_one_d = tf.constant([2,0,-1], dtype = tf.float32)
casted_tensor_one_d = tf.cast(tensor_one_d, dtype = tf.int64)
bool_tensor_one_d = tf.cast(tensor_one_d, dtype = tf.bool)
print(tensor_one_d)
print(casted_tensor_one_d)
print(bool_tensor_one_d)
```

```python
tensor_bool = tf.constant([True, False, True])
print(tensor_bool)
```

```python
tensor_two_d = tf.constant([[1,2,3],
                             [2,3,4],
                             [5,6,7]])
print(tensor_two_d)
```

```python
tensor_three_d = tf.constant([[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]])
print(tensor_three_d.shape)
tensor_three_d_2 = tf.constant([[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]])
print(tensor_three_d_2.ndim)
```

```python
print(tensor_three_d.ndim) # Dimension of our tensor
```

# Indexing Dim. > 2

```python
tensor_two_d = tf.constant([[1,2,3],
                             [2,3,4],
                             [5,6,7]])

print(tensor_two_d[0:3,0:2]) #for rows, for columns
print(tensor_two_d[0:3,:]) # all columns
print(tensor_two_d[2]) #the whole of the third row
print(tensor_two_d[:,1])
print(tensor_two_d[...,1]) #all rows of the 2nd column 
```

```python
tensor_three_d = tf.constant([[[1,2,3],
                             [2,3,4],
                             [5,6,7]],

                             [[1,2,3],
                             [2,3,4],
                             [5,6,7]]])

print(tensor_three_d[0,:,:])
print(tensor_three_d[0,0,:])
print(tensor_three_d[0,:,-1])
print(tensor_three_d[0:2,:,-1])
print(tensor_three_d[...,:,1])

```

# Math Operations

```python
x = tf.constant([-2.25, 3.25])
tf.abs(x)
```

```python
# complex number
x_abs_complex = tf.constant([[-2.25 + 4.75j], [-3.25 + 5.75j]])
tf.abs(x_abs_complex)
```

```python
x = tf.constant([1, 2], dtype=tf.int8)
y = tf.constant([3, 0], dtype=tf.int8)
print(tf.add(x,y))
print(tf.multiply(x,y))
print(tf.subtract(x,y))
print(tf.math.divide_no_nan(x,y)) #gives 0. when divide by 0
```

```python
x2 = tf.constant([[1, 2], [3,4]], dtype=tf.int8)
x3 = x2 = tf.constant([1, 2], dtype=tf.int8)
y2 = tf.constant([3], dtype=tf.int8) #this tensor is broadcasted
y3 = tf.constant([[3], [6], [9]], dtype=tf.int8)
print(tf.add(x2,y2))
print(tf.math.multiply(x2, y3))
```
